```python
import os
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
from PIL import Image
import torch
import torch.nn as nn
import torchvision.models.segmentation as segmentation
import numpy as np
import sys
import torchvision.models as models


model = segmentation.deeplabv3_resnet50(pretrained=False, num_classes=8)
classifier = list(model.classifier.children())
classifier.append(nn.Softmax(dim=1))
model.classifier = nn.Sequential(*classifier)

model.load_state_dict(torch.load('/home/amit/Downloads/deeplab_modelsoftmax.pth'))

model.eval()
```

```python
transform = transforms.Compose([
    transforms.Resize((256,256)),
    transforms.ToTensor(),
])

def preprocess_image(image_path):
    image = Image.open(image_path).convert("RGB")
    image = transform(image)
    return image
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
image_path = "/home/amit/leftImg8bit_trainvaltest/leftImg8bit/test/berlin/berlin_000000_000019_leftImg8bit.png"
image = preprocess_image(image_path).unsqueeze(0)
image = image.to(device)
model = model.to(device)
with torch.no_grad():
    output = model(image)
```
